```python
# Descargamos datos FRED
df_d = pd.DataFrame(fred.get_series('DDFUELUSGULF','2022-01-01'))
df_d = df_d.dropna()
df_d.index.names = ['Date']
df_d.columns = ["diesel_int"]
df_d
```

```python
# Descargamos datos de la acción sleccionada:
df_trm = yfin.download('COP=X', start='2022-01-01')
#df_trm = df_trm.dropna()
df_trm.index.names = ['Date']
# Drop multiple columns
df_trm = df_trm.drop(['Open', 'High', 'Low', 'Close', 'Volume'], axis=1)
df_trm.columns = ["TRM"]
# Fill NaN values in column 'rf' with the lag
df_trm['TRM'] = df_trm['TRM'].fillna(method='ffill')
df_trm
```

```python
# Merge the DataFrames on the index
df = pd.merge(df_d, df_trm, on='Date')
df
```
